```python
# Convert prediction strings/lists to actual Python lists
pred_lists = data["prediction"].apply(
    lambda x: x if isinstance(x, list) else eval(x)
)

# Expand into columns: pred_0 ... pred_11
pred_df = pd.DataFrame(pred_lists.tolist(), index=data.index)
pred_df.columns = [f"pred_{i}" for i in range(pred_df.shape[1])]

# Range (max - min) at each corresponding prediction index across all molecules
index_ranges = pred_df.max(axis=0) - pred_df.min(axis=0)

# Optional: include min and max for clarity
range_summary = pd.DataFrame({
    "min": pred_df.min(axis=0),
    "max": pred_df.max(axis=0),
    "range": index_ranges
})

range_summary
```
