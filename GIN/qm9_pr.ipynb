## Extracting Coordinates 

```python
c = data["df_atoms"][0]
c = c.split("\n")
c = c[1:-1]
c
```

```python
atom_rows = []
for line in c:
    parts = line.split()
    if len(parts) >= 6:
        atom_rows.append(parts[1:6])  # atom, x, y, z, charge

atom_df = pd.DataFrame(atom_rows, columns=["atom", "x", "y", "z", "charge"])
atom_df[["x", "y", "z", "charge"]] = atom_df[["x", "y", "z", "charge"]].astype(float)
atom_df
```
