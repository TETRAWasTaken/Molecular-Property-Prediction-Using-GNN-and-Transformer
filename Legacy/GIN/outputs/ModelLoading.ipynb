```python


def read_extended_xyz(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        
    # 1. Get the number of atoms (Line 1)
    num_atoms = int(lines[0].strip())
    
    # 2. Get the molecule properties/metadata (Line 2)
    properties = lines[1].strip().split()
    
    # 4. Extract the remaining footer metadata
    current_line = 2 + num_atoms
    
    frequencies = lines[current_line].strip().split()
    smiles = lines[current_line + 1].strip()
    
    # Bundle everything into a dictionary
    molecule_data = {
        'num_atoms': num_atoms,
        'properties': properties,
        'frequencies': [float(f) for f in frequencies],
        'smiles': smiles,
    }
    
    return molecule_data
```
